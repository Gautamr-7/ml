# .DROP() 

```python
import pandas as pd
# reads csv file
df = pd.read_csv('players_20.csv')
# set index
df.set_index('short_name', inplace=True)
# select column
df = df[['long_name', 'age', 'dob', 'height_cm', 'weight_kg', 'nationality', 'club']]
```

```python
df.drop(['L. Messi'],axis=0,inplace=True)
```

```python
df.drop(['club'],axis=1)
```

# .sample()

```python
df['nationality'].sample(5)
```

```python
df['nationality'].sample(5, random_state=99)
#99- least change in sample all the way to 1
```

```python
df.shape
```

```python
df['nationality'].sample(frac=.1, random_state=99).shape
#99- least change in sample all the way to 1            #FRAC=.1 MEANS 10% OF DF TO GIVE AS SAMPLE
```

# REPLACE 2X

```python
# upsample: increase the sampling rate -> frac>1  (Note: replace parameter has to be True for frac parameter > 1)
df.sample(frac=2, replace=True, random_state=99)
```
