# using np.where()

```python
import numpy as np
df['Price Tier']=np.where(df['Price']>2000,'Expensive','cheap')
df.head(2)
df['Price Tier'].value_counts()
```

```python
df.shape
```

# using np.select()

```python
conditions = [
    (df['Price']>3000),
    (df['Price']>2000) & (df['Price']<=3000),
    (df['Price']>800) & (df['Price']<=2000),
    (df['Price']<=800),
]
choices = ['Too Expensive', 'Expensive', 'Affordable', 'Cheap']
```

```python
df['select_value']=np.select(conditions,choices,default='other')
```

```python
df.head(3)
df['select_value'].value_counts()
```
