```python


import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Load built-in datasets (no file needed!)
tips    = sns.load_dataset("tips")      # restaurant tips
titanic = sns.load_dataset("titanic")   # survival data
iris    = sns.load_dataset("iris")      # flower measurements
```

```python
# 5 styles: darkgrid, whitegrid, dark, white, ticks
sns.set_style("whitegrid")  # best for ML reports

# 4 contexts: paper, notebook, talk, poster
sns.set_context("notebook", font_scale=1.3)

# Color palettes
sns.set_palette("husl")  # or "muted", "deep", "colorblind"
```

# plot types


# 1. Distribution Plots — "What does my data look like?"

```python
# Histogram + KDE (replaces old distplot)
fig, axes = plt.subplots(1, 2, figsize=(10, 3))

# Just histogram
sns.histplot(data=tips, x="total_bill", bins=20, ax=axes[0])
axes[0].set_title("Histogram")

# Histogram + KDE overlay
sns.histplot(data=tips, x="total_bill", kde=True, ax=axes[1])
axes[1].set_title("Histogram + KDE")
plt.show()
```

```python
# Separate distributions by category
sns.histplot(data=tips, x="total_bill", hue="sex", kde=True, alpha=0.5)
plt.show()
```

```python
# Basic scatter
sns.scatterplot(data=tips, x="total_bill", y="tip")

# With color (3rd dimension!) and size (4th dimension!)
sns.scatterplot(data=tips, x="total_bill", y="tip",
                hue="sex", size="size", style="smoker")
plt.show()
```
