# DATA EXTRACTION

```python
import pandas as pd
df=pd.read_csv('population_total.csv')
df.head(3)
#df.nunique()['country']
```

```python
#drop na values
df.dropna( inplace=True)



df_pivot=df.pivot(index='year',columns='country',values='population')
```

# LINE GRAPH

```python
df_pivot.plot(kind='line',
              xlabel='Years', ylabel='Population',
              title='Population (1955-2020)',
              figsize=(8,4))
```

# BAR GRAPH

```python
df_pivot[['China','India']].plot(kind='bar',color='green',xlabel='country',ylabel='year',title='pop %')
```

```python
# selecting a few years
df_pivot_sample = df_pivot[df_pivot.index.isin([1980, 1990, 2000, 2010, 2020])]
# making grouped barplot
df_pivot_sample.plot(kind='bar',
                     xlabel='Years', ylabel='Population',
                     figsize=(8,4))
```

# PIECHART

```python
df_pivot=df_pivot[['China','India','United States']]
```

```python
# 1. Select the data for the year 2020 from your pivoted dataframe
# Using [[2020]] keeps it as a DataFrame so .plot() works easily
df_2020 = df_pivot.loc[[2020]]

# 2. Transpose it (swap rows and columns) 
# Pie charts need the categories to be in the index
df_2020_t = df_2020.transpose()

# 3. Plot using the year column
# Use the actual column name (2020)
df_2020_t.plot(kind='pie', y=2020, title='Population in 2020 (%)', autopct='%1.1f%%')
```
